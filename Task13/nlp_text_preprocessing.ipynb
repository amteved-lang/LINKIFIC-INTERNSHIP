```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

df = pd.read_csv("text_dataset.csv")

print("===== NLP TEXT PREPROCESSING =====")

print("\n===== ORIGINAL DATASET =====")
print(df)

def preprocess_text(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", "", text)
    tokens = word_tokenize(text)
    stop_words = set(stopwords.words("english"))
    tokens = [word for word in tokens if word not in stop_words]
    return " ".join(tokens)

df["Processed_Text"] = df["Text"].apply(preprocess_text)

print("\n===== PREPROCESSED TEXT =====")
print(df[["Text", "Processed_Text"]])

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df["Processed_Text"])

tfidf_output = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print("\n===== TF-IDF FEATURE NAMES =====")
print(vectorizer.get_feature_names_out())

print("\n===== TF-IDF OUTPUT =====")
print(tfidf_output.round(3))

print("\n===== TF-IDF MATRIX SHAPE =====")
print(tfidf_matrix.shape)

tfidf_output.to_csv("tfidf_output.csv", index=False)

print("\n===== NLP SUMMARY =====")
print("Total Documents:", len(df))
print("Total TF-IDF Features:", len(vectorizer.get_feature_names_out()))
print("TF-IDF output saved as tfidf_output.csv")
print("Text preprocessing and TF-IDF vectorization completed successfully.")
```
