```python
import numpy as np

print("NumPy version:", np.__version__)
```

```python
# 1D NumPy Array

marks = np.array([78, 85, 92, 67, 88])

print("1D Array:")
print(marks)

print("Shape:", marks.shape)
print("Data Type:", marks.dtype)
```

```python
# 2D NumPy Array

student_marks = np.array([
    [78, 85, 92],
    [67, 88, 76],
    [90, 95, 89]
])

print("2D Array:")
print(student_marks)

print("Shape:", student_marks.shape)
```

```python
# NumPy Indexing

print("First element:", marks[0])
print("Second element:", marks[1])
print("Last element:", marks[-1])

print("First student, first subject:", student_marks[0, 0])
print("Second student, third subject:", student_marks[1, 2])
```

```python
# NumPy Slicing

print("First three marks:")
print(marks[0:3])

print("\nLast three marks:")
print(marks[2:5])

print("\nFirst two students:")
print(student_marks[0:2])

print("\nFirst two subjects:")
print(student_marks[:, 0:2])
```

```python
# Mathematical Operations

print("Marks:", marks)

print("Sum:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
```

```python
# Mathematical Operations on 2D Array

print("Student Marks:")
print(student_marks)

print("\nTotal of all marks:", np.sum(student_marks))
print("Average of all marks:", np.mean(student_marks))
print("Highest mark:", np.max(student_marks))
print("Lowest mark:", np.min(student_marks))
```

```python
# Python List vs NumPy Array

python_list = [10, 20, 30, 40]

numpy_array = np.array([10, 20, 30, 40])

print("Python List:")
print(python_list)

print("\nNumPy Array:")
print(numpy_array)

# Python list multiplication
print("\nPython List * 2:")
print(python_list * 2)

# NumPy array multiplication
print("\nNumPy Array * 2:")
print(numpy_array * 2)
```
