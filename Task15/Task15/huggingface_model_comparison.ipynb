```python
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM

print("===== HUGGING FACE MODEL COMPARISON =====")

text_generation = pipeline(
    "text-generation",
    model="distilgpt2"
)

sentiment_analysis = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

generation_prompt = "Artificial intelligence is transforming the way people"

sentiment_text = "I really enjoyed learning about machine learning and artificial intelligence."

summary_text = """
Artificial intelligence has developed significantly over the past few decades.
Traditional machine learning systems were designed to learn patterns from data
and make predictions. Deep learning introduced neural networks with many layers,
which improved performance in areas such as computer vision and speech recognition.
Generative AI and Large Language Models have expanded these capabilities by
allowing systems to generate text, summarize information, answer questions,
write code, and interact with users using natural language.
"""

print("\n===== TEXT GENERATION =====")
generation_result = text_generation(
    generation_prompt,
    max_new_tokens=50,
    num_return_sequences=1
)
print(generation_result[0]["generated_text"])

print("\n===== SENTIMENT ANALYSIS =====")
sentiment_result = sentiment_analysis(sentiment_text)
print(sentiment_result)

print("\n===== TEXT SUMMARIZATION =====")

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

summary_prompt = "Summarize the following text: " + summary_text

inputs = tokenizer(
    summary_prompt,
    return_tensors="pt",
    truncation=True,
    max_length=512
)

outputs = model.generate(
    **inputs,
    max_new_tokens=80
)

summary = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(summary)

print("\n===== MODEL COMPARISON =====")
print("Text Generation Model: distilgpt2")
print("Purpose: Generates text based on a given prompt.")

print("\nSentiment Analysis Model: distilbert-base-uncased-finetuned-sst-2-english")
print("Purpose: Classifies text sentiment as positive or negative.")

print("\nSummarization Model: google/flan-t5-small")
print("Purpose: Generates a concise summary of provided text.")

print("\n===== TASK COMPLETED =====")
print("Three Hugging Face models were successfully tested.")
```
