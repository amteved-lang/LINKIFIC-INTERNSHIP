# 🧹 Data Cleaning and Preprocessing

## Project Overview

This project demonstrates basic data cleaning and preprocessing techniques using Python and Pandas.

The dataset contains student information and academic performance data. The objective is to identify and handle missing values, remove duplicate records, rename columns, convert data types, and save the cleaned dataset.

## Objectives

- Load a dataset using Pandas
- Identify missing values
- Handle missing values
- Remove duplicate records
- Rename columns
- Convert incorrect data types
- Save the cleaned dataset

## Technologies Used

- Python
- Pandas
- Jupyter Notebook
- VS Code

```python
import pandas as pd
import os

print("Current folder:", os.getcwd())
print("Files in folder:", os.listdir())

df = pd.read_csv("dirty_students.csv")

print("\n===== ORIGINAL DATASET =====")
print(df)

print("\n===== DATASET SHAPE =====")
print(df.shape)

print("\n===== DATA TYPES BEFORE CLEANING =====")
print(df.dtypes)

print("\n===== MISSING VALUES =====")
print(df.isnull().sum())

print("\n===== DUPLICATE RECORDS =====")
print(df.duplicated().sum())

df.columns = ["Name", "Age", "Gender", "Python", "Math", "Science", "Attendance", "City"]

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Python"] = pd.to_numeric(df["Python"], errors="coerce")
df["Math"] = pd.to_numeric(df["Math"], errors="coerce")
df["Science"] = pd.to_numeric(df["Science"], errors="coerce")
df["Attendance"] = pd.to_numeric(df["Attendance"], errors="coerce")

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Python"] = df["Python"].fillna(df["Python"].median())
df["Math"] = df["Math"].fillna(df["Math"].median())
df["Science"] = df["Science"].fillna(df["Science"].median())
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].median())

df["Name"] = df["Name"].str.strip()
df["Gender"] = df["Gender"].str.strip().str.title()
df["City"] = df["City"].str.strip().str.title()

df = df.drop_duplicates()

df["Age"] = df["Age"].astype(int)
df["Python"] = df["Python"].astype(float)
df["Math"] = df["Math"].astype(float)
df["Science"] = df["Science"].astype(float)
df["Attendance"] = df["Attendance"].astype(float)

print("\n===== CLEANED DATASET =====")
print(df)

print("\n===== MISSING VALUES AFTER CLEANING =====")
print(df.isnull().sum())

print("\n===== DUPLICATES AFTER CLEANING =====")
print(df.duplicated().sum())

print("\n===== DATA TYPES AFTER CLEANING =====")
print(df.dtypes)

print("\n===== CLEANED DATASET SHAPE =====")
print(df.shape)

df.to_csv("cleaned_students.csv", index=False)

print("\n===== PROJECT COMPLETED =====")
print("cleaned_students.csv has been created successfully.")
```

## 🧹 Data Cleaning Summary

The dataset was successfully cleaned using Pandas.

### Cleaning Operations Performed

- Identified missing values using `isnull()`
- Filled missing numerical values using median values
- Renamed columns to clear and standardized names
- Converted numerical columns to appropriate data types
- Removed duplicate records using `drop_duplicates()`
- Standardized text values for names, gender, and city
- Saved the final cleaned dataset as `cleaned_students.csv`

### Result

The final dataset contains no missing values or duplicate records and is ready for further analysis or machine learning.
