```python
import pandas as pd

df = pd.read_csv("students.csv")

print("Pandas version:", pd.__version__)
print("\nDataset:")
print(df)

print("\nFirst 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

print("\nNumber of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataFrame Information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

print("\nStudents with Python marks above 80:")
print(df[df["Python"] > 80])

print("\nStudents with Math marks above 85:")
print(df[df["Math"] > 85])

print("\nStudents with Python and Math marks above 80:")
print(df[(df["Python"] > 80) & (df["Math"] > 80)])

print("\nStudents with Python or Science marks above 90:")
print(df[(df["Python"] > 90) | (df["Science"] > 90)])

print("\nStudents from Nagpur:")
print(df[df["City"] == "Nagpur"])

print("\nStudents with attendance above 90%:")
print(df[df["Attendance"] > 90])

print("\nSummary Statistics:")
print(df.describe())

print("\nSubject Statistics:")
print("Python Average:", round(df["Python"].mean(), 2))
print("Python Maximum:", df["Python"].max())
print("Python Minimum:", df["Python"].min())
print("Math Average:", round(df["Math"].mean(), 2))
print("Math Maximum:", df["Math"].max())
print("Math Minimum:", df["Math"].min())
print("Science Average:", round(df["Science"].mean(), 2))
print("Science Maximum:", df["Science"].max())
print("Science Minimum:", df["Science"].min())

df["Total_Marks"] = df["Python"] + df["Math"] + df["Science"]
df["Average_Marks"] = df["Total_Marks"] / 3

print("\nStudent Total and Average Marks:")
print(df[["Name", "Python", "Math", "Science", "Total_Marks", "Average_Marks"]])

top_student_index = df["Average_Marks"].idxmax()
top_student = df.loc[top_student_index]

print("\nTop Student:", top_student["Name"])
print("Top Student Average:", round(top_student["Average_Marks"], 2))

print("\nStudents with Average Marks above 80:")
print(df[df["Average_Marks"] > 80][["Name", "Average_Marks"]])

print("\nFinal Analysis:")
print("Total Students:", len(df))
print("Total Columns:", len(df.columns))
print("Overall Python Average:", round(df["Python"].mean(), 2))
print("Overall Math Average:", round(df["Math"].mean(), 2))
print("Overall Science Average:", round(df["Science"].mean(), 2))
print("Overall Attendance:", round(df["Attendance"].mean(), 2), "%")
print("Highest Average:", round(df["Average_Marks"].max(), 2))
print("Top Student:", top_student["Name"])
print("Students with Average > 80:", len(df[df["Average_Marks"] > 80]))
```
